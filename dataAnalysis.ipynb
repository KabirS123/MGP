```python
import pandas
import numpy
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
data = pandas.read_csv("01_health_and_income.csv")
columnNames = ["Life expectancy (years)", "GDP per capita (2021 international-$)"]
aggregateStatisticalSummaries = data[columnNames].describe().T.drop(columns=["count"]).rename(columns={"mean": "Mean", "std": "Standard Deviation", "min": "Minimum", "25%": "Lower Quartile", "50%": "Median", "75%": "Upper Quartile", "max": "Maximum"})

aggregateStatisticalSummaries["Interquartile Range"] = aggregateStatisticalSummaries["Upper Quartile"] - aggregateStatisticalSummaries["Lower Quartile"]
aggregateStatisticalSummaries["Range"] = aggregateStatisticalSummaries["Maximum"] - aggregateStatisticalSummaries["Minimum"]
aggregateStatisticalSummaries 

#This displays a table showing statistical summaries for the whole data set. The table includes the mean, standard deviation, minimum, lower quartile, median, upper quartile, maximum, interquartile range, and range for each of the two columns: "Life expectancy (years)" and "GDP per capita (2021 international-$)".
```
