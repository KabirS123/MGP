```python
import pandas
import numpy
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from IPython.display import display
```

```python
data = pandas.read_csv("01_health_and_income.csv")
columnNames = ["Life expectancy (years)", "GDP per capita (2021 international-$)"]
richest = data[data["Year"] == 2023].nlargest(10, "GDP per capita (2021 international-$)")  
poorest = data[data["Year"] == 2023].nsmallest(10, "GDP per capita (2021 international-$)")
richest = richest["Entity"].to_list()
poorest = poorest["Entity"].to_list()

columns = {
        "mean": "Mean",
        "std": "Standard Deviation",
        "min": "Minimum",
        "25%": "Lower Quartile",
        "50%": "Median",
        "75%": "Upper Quartile",
        "max": "Maximum"
    }
```

```python

aggregateStatisticalSummaries = data[columnNames].describe().T.drop(columns=["count"]).rename(columns={"mean": "Mean", "std": "Standard Deviation", "min": "Minimum", "25%": "Lower Quartile", "50%": "Median", "75%": "Upper Quartile", "max": "Maximum"})

aggregateStatisticalSummaries["Interquartile Range"] = aggregateStatisticalSummaries["Upper Quartile"] - aggregateStatisticalSummaries["Lower Quartile"]
aggregateStatisticalSummaries["Range"] = aggregateStatisticalSummaries["Maximum"] - aggregateStatisticalSummaries["Minimum"]
aggregateStatisticalSummaries 

#This displays a table showing statistical summaries for the whole data set. The table includes the mean, standard deviation, minimum, lower quartile, median, upper quartile, maximum, interquartile range, and range for each of the two columns: "Life expectancy (years)" and "GDP per capita (2021 international-$)".
```

```python
# Select 2008 observations for the countries listed in richest
richData2008 = data[(data["Entity"].isin(richest)) & (data["Year"] == 2008)]
richSummaries2008 = (richData2008[columnNames].describe().T.drop(columns=["count"]).rename(columns=columns))
richSummaries2008["Interquartile Range"] = (richSummaries2008["Upper Quartile"] - richSummaries2008["Lower Quartile"])
richSummaries2008["Range"] = (richSummaries2008["Maximum"] - richSummaries2008["Minimum"])

poorData2008 = data[(data["Entity"].isin(poorest)) & (data["Year"] == 2008)]
poorSummaries2008 = (poorData2008[columnNames].describe().T.drop(columns=["count"]).rename(columns=columns))
poorSummaries2008["Interquartile Range"] = (poorSummaries2008["Upper Quartile"] - poorSummaries2008["Lower Quartile"])
poorSummaries2008["Range"] = (poorSummaries2008["Maximum"] - poorSummaries2008["Minimum"])


display(richSummaries2008)
display(poorSummaries2008)
```
