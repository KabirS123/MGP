```python
import pandas
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from IPython.display import display, Markdown
from scipy.stats import linregress as lr
```

```python
data = pandas.read_csv("01_health_and_income.csv")
columnNames = ["Life expectancy (years)", "GDP per capita (2021 international-$)"]
richest = data[data["Year"] == 2023].nlargest(10, "GDP per capita (2021 international-$)")  
poorest = data[data["Year"] == 2023].nsmallest(10, "GDP per capita (2021 international-$)")
richest = richest["Entity"].to_list()
poorest = poorest["Entity"].to_list()

columns = {
        "mean": "Mean",
        "std": "Standard Deviation",
        "min": "Minimum",
        "25%": "Lower Quartile",
        "50%": "Median",
        "75%": "Upper Quartile",
        "max": "Maximum"
    }
```

```python
aggregateData = data[(data["Entity"].isin(poorest) | data["Entity"].isin(richest)) & (data["Year"] >= 2008) & (data["Year"] <= 2023)]

aggregateStatisticalSummaries = aggregateData[columnNames].describe().T.drop(columns=["count"]).rename(columns={"mean": "Mean", "std": "Standard Deviation", "min": "Minimum", "25%": "Lower Quartile", "50%": "Median", "75%": "Upper Quartile", "max": "Maximum"})

aggregateStatisticalSummaries["Interquartile Range"] = aggregateStatisticalSummaries["Upper Quartile"] - aggregateStatisticalSummaries["Lower Quartile"]
aggregateStatisticalSummaries["Range"] = aggregateStatisticalSummaries["Maximum"] - aggregateStatisticalSummaries["Minimum"]

display(Markdown("### Aggregate Statistical Summaries for the Richest and Poorest Countries (2008-2023)"))
display(aggregateStatisticalSummaries) 

#This displays a table showing statistical summaries for the richest and poorest countries for our period of 2008-2023 inclusive. The table includes the mean, standard deviation, minimum, lower quartile, median, upper quartile, maximum, interquartile range, and range for each of the two columns: "Life expectancy (years)" and "GDP per capita (2021 international-$)".
```

# Explanation:

The GDP chart for rich countries experiences a significant fall mainly due to one territory, Macau. This is because a large portion of Macau's GDP came from gambling tourism, therefore, when the pandemic started, Macau saw a significant fall in tourism and therefore GDP (having fallen from 133,454 to 59,662 )

Furthermore, whilst at first the GDP chart may appear like the richer countries (blue) are doing worse from 2008 onwards, the graph is showing both groups indexed at their 2008 values. Therefore, when a line moves up or down, this is a percentage change from its previous year rather than an absolute change (and must be accounted for)

```python
gdpCol = "GDP per capita (2021 international-$)"
leCol = "Life expectancy (years)"

# Work on a copy so the original DataFrame is untouched
plotData = totalData2023.copy()
plotData["Log GDP"] = np.log(plotData[gdpCol])
plotData["Group"] = np.where(plotData["Entity"].isin(richest), "Richest 10", "Poorest 10")

# Pooled fit on log GDP
pooled = lr(plotData["Log GDP"], plotData[leCol])

fig, ax = plt.subplots(figsize=(10, 6))

for group, colour in [("Poorest 10", "tab:red"), ("Richest 10", "tab:blue")]:
    sub = plotData[plotData["Group"] == group]
    ax.scatter(sub["Log GDP"], sub[leCol], color=colour, label=group, s=45)
    for _, row in sub.iterrows():
        ax.annotate(row["Entity"], (row["Log GDP"], row[leCol]),
                    xytext=(4, 4), textcoords="offset points", fontsize=7)

# Pooled regression line
xs = np.linspace(plotData["Log GDP"].min(), plotData["Log GDP"].max(), 100)
ax.plot(xs, pooled.intercept + pooled.slope * xs, color="red", linewidth=2, label="Pooled fit")

# Log-scale tick labels shown in dollars
ticks = [1000, 2000, 5000, 20000, 50000, 100000]
ax.set_xticks(np.log(ticks))
ax.set_xticklabels([f"{t:,}" for t in ticks])

ax.set_xlabel("GDP per person, 2021 international-$ (logarithmic)")
ax.set_ylabel("Life expectancy (years)")
ax.set_title("GDP per person vs life expectancy, 2023\n10 richest and 10 poorest countries")

perDoubling = pooled.slope * np.log(2)
ax.text(0.03, 0.95,
        f"Pooled: y = {pooled.slope:.2f}·ln(GDP) + {pooled.intercept:.1f}\n"
        f"R² = {pooled.rvalue**2:.2f}, n = {len(plotData)}\n"
        f"≈ {perDoubling:.1f} years per doubling of GDP",
        transform=ax.transAxes, va="top",
        bbox=dict(boxstyle="round", facecolor="white", alpha=0.8))

ax.legend(loc="lower right")
plt.tight_layout()
plt.show()

# Within-group fits for the counter-evidence section
for group in ["Poorest 10", "Richest 10"]:
    sub = plotData[plotData["Group"] == group]
    fit = lr(sub["Log GDP"], sub[leCol])
    print(f"{group}: slope = {fit.slope:.2f}, R² = {fit.rvalue**2:.2f}, p = {fit.pvalue:.3f}, n = {len(sub)}")
```

```python
gdpCol = "GDP per capita (2021 international-$)"
leCol = "Life expectancy (years)"

def regressionPanel(ax, df, title, ticks):
    df = df.dropna(subset=[gdpCol, leCol]).copy()
    df["Log GDP"] = np.log(df[gdpCol])
    fit = lr(df["Log GDP"], df[leCol])

    for country, sub in df.groupby("Entity"):
        ax.scatter(sub["Log GDP"], sub[leCol], s=18, alpha=0.7, label=country)

    xs = np.linspace(df["Log GDP"].min(), df["Log GDP"].max(), 100)
    ax.plot(xs, fit.intercept + fit.slope * xs, color="black", linewidth=2)

    ax.set_xticks(np.log(ticks))
    ax.set_xticklabels([f"{t:,}" for t in ticks], rotation=30)
    ax.set_xlabel("GDP per person, 2021 int-$ (log scale)")
    ax.set_ylabel("Life expectancy (years)")
    ax.set_title(title)
    ax.text(0.03, 0.97,
            f"y = {fit.slope:.2f}·ln(GDP) + {fit.intercept:.1f}\n"
            f"R² = {fit.rvalue**2:.2f}, p = {fit.pvalue:.1e}\n"
            f"n = {len(df)} country-years",
            transform=ax.transAxes, va="top",
            bbox=dict(boxstyle="round", facecolor="white", alpha=0.85))
    ax.legend(fontsize=6, ncol=2, loc="lower right")
    return fit

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
poorFit = regressionPanel(axes[0], aggregatePoorData, "Poorest 10 countries, 2008-2023", [1000, 1500, 2000, 3000])
richFit = regressionPanel(axes[1], aggregateRichData, "Richest 10 countries, 2008-2023", [60000, 80000, 100000, 130000])
fig.suptitle("GDP per person vs life expectancy within each group (country-years pooled)")
plt.tight_layout()
plt.show()
```
