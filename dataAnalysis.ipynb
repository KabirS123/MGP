```python
import pandas
import numpy
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from IPython.display import display, Markdown
```

```python
data = pandas.read_csv("01_health_and_income.csv")
columnNames = ["Life expectancy (years)", "GDP per capita (2021 international-$)"]
richest = data[data["Year"] == 2023].nlargest(10, "GDP per capita (2021 international-$)")  
poorest = data[data["Year"] == 2023].nsmallest(10, "GDP per capita (2021 international-$)")
richest = richest["Entity"].to_list()
poorest = poorest["Entity"].to_list()

columns = {
        "mean": "Mean",
        "std": "Standard Deviation",
        "min": "Minimum",
        "25%": "Lower Quartile",
        "50%": "Median",
        "75%": "Upper Quartile",
        "max": "Maximum"
    }
```

```python
aggregateData = data[(data["Entity"].isin(poorest) | data["Entity"].isin(richest)) & (data["Year"] >= 2008) & (data["Year"] <= 2023)]

aggregateStatisticalSummaries = aggregateData[columnNames].describe().T.drop(columns=["count"]).rename(columns={"mean": "Mean", "std": "Standard Deviation", "min": "Minimum", "25%": "Lower Quartile", "50%": "Median", "75%": "Upper Quartile", "max": "Maximum"})

aggregateStatisticalSummaries["Interquartile Range"] = aggregateStatisticalSummaries["Upper Quartile"] - aggregateStatisticalSummaries["Lower Quartile"]
aggregateStatisticalSummaries["Range"] = aggregateStatisticalSummaries["Maximum"] - aggregateStatisticalSummaries["Minimum"]

display(Markdown("### Aggregate Statistical Summaries for the Richest and Poorest Countries (2008-2023)"))
display(aggregateStatisticalSummaries) 

#This displays a table showing statistical summaries for the richest and poorest countries for our period of 2008-2023 inclusive. The table includes the mean, standard deviation, minimum, lower quartile, median, upper quartile, maximum, interquartile range, and range for each of the two columns: "Life expectancy (years)" and "GDP per capita (2021 international-$)".
```

```python
# Select 2008 observations for the countries listed in richest
richData2008 = data[(data["Entity"].isin(richest)) & (data["Year"] == 2008)]
richSummaries2008 = (richData2008[columnNames].describe().T.drop(columns=["count"]).rename(columns=columns))
richSummaries2008["Interquartile Range"] = (richSummaries2008["Upper Quartile"] - richSummaries2008["Lower Quartile"])
richSummaries2008["Range"] = (richSummaries2008["Maximum"] - richSummaries2008["Minimum"])

poorData2008 = data[(data["Entity"].isin(poorest)) & (data["Year"] == 2008)]
poorSummaries2008 = (poorData2008[columnNames].describe().T.drop(columns=["count"]).rename(columns=columns))
poorSummaries2008["Interquartile Range"] = (poorSummaries2008["Upper Quartile"] - poorSummaries2008["Lower Quartile"])
poorSummaries2008["Range"] = (poorSummaries2008["Maximum"] - poorSummaries2008["Minimum"])

totalData2008 = data[((data["Entity"].isin(poorest)) | (data["Entity"].isin(richest))) & (data["Year"] == 2008)]
totalSummaries2008 = (totalData2008[columnNames].describe().T.drop(columns=["count"]).rename(columns=columns))
totalSummaries2008["Interquartile Range"] = (totalSummaries2008["Upper Quartile"] - totalSummaries2008["Lower Quartile"])
totalSummaries2008["Range"] = (totalSummaries2008["Maximum"] - totalSummaries2008["Minimum"])

display(Markdown("### Richest countries - 2008"))
display(richSummaries2008)
display(Markdown("### Poorest countries - 2008"))
display(poorSummaries2008)
display(Markdown("### All countries - 2008"))
display(totalSummaries2008)
```
